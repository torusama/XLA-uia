# BTXRD ViT: small-lesion Dice A/B
Same checkpoint, batch-4 paired augmentation, 20 epochs and batch Dice + weighted BCE. Candidate adds 0.5 times mean soft Dice loss for augmented masks with 1-255 pixels; control uses the original loss. Empty and larger masks do not enter the extra term.
Validation uses full images; test is not evaluated. Compare foreground micro Dice, small-lesion Dice and normal false positives; do not compare losses across arms.
